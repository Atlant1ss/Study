```python
#exercício 1
def valorPagamento(valorPrestacao, diasAtraso):

    if diasAtraso <= 0:
        return valorPrestacao

    multa  = valorPrestacao * 0.03
    juros  = valorPrestacao * 0.001 * diasAtraso
    return valorPrestacao + multa + juros

def main():
    quantidade = 0
    totalDia   = 0.0

    print("SISTEMA DE PAGAMENTO DE PRESTAÇÕES")
    print("Digite 0 (zero) para encerrar e ver o relatório.\n")

    while True:
        while True:
            entrada = input("Valor da prestação (R$): ").strip().replace(",", ".")
            try:
                prestacao = float(entrada)
                if prestacao < 0:
                    print("  Valor inválido. Digite um número maior ou igual a zero.\n")
                    continue
                break
            except ValueError:
                print("  Entrada inválida. Digite apenas números.\n")

        if prestacao == 0:
            break

        while True:
            entrada_dias = input("Dias em atraso: ").strip()
            try:
                dias = int(entrada_dias)
                if dias < 0:
                    print("  Valor inválido. Digite um número inteiro maior ou igual a zero.\n")
                    continue
                break
            except ValueError:
                print("  Entrada inválida. Digite apenas números inteiros.\n")

        total = valorPagamento(prestacao, dias)

        if dias == 0:
            print(f"  Valor a pagar: R$ {total:.2f}  (sem atraso)")
        else:
            multa = prestacao * 0.03
            juros = prestacao * 0.001 * dias
            print(f"  Prestação original: R$ {prestacao:.2f}")
            print(f"  Multa (3%):         R$ {multa:.2f}")
            print(f"  Juros (0,1%/dia):   R$ {juros:.2f}  ({dias} dia(s))")
            print(f"  Valor a pagar:      R$ {total:.2f}")

        quantidade += 1
        totalDia   += total

    print("           RELATÓRIO DO DIA")
    print(f"  Prestações pagas:  {quantidade}")
    print(f"  Total arrecadado:  R$ {totalDia:.2f}")
main()
```

```python
#exercício 2
def criar_tabuleiro():
    return [[' ' for _ in range(3)] for _ in range(3)]


def exibir_tabuleiro(tabuleiro):
    print("\n    Col 1   Col 2   Col 3")
    for i, linha in enumerate(tabuleiro):
        print(f"  |       |       |       |")
        print(f"  |  {linha[0]}    |  {linha[1]}    |  {linha[2]}    |  ")
        print(f"  |       |       |       |")

def verificar_vencedor(tabuleiro):
    for linha in tabuleiro:
        if linha[0] == linha[1] == linha[2] != ' ':
            return linha[0]

    for col in range(3):
        if tabuleiro[0][col] == tabuleiro[1][col] == tabuleiro[2][col] != ' ':
            return tabuleiro[0][col]


    if tabuleiro[0][0] == tabuleiro[1][1] == tabuleiro[2][2] != ' ':
        return tabuleiro[0][0]
    if tabuleiro[0][2] == tabuleiro[1][1] == tabuleiro[2][0] != ' ':
        return tabuleiro[0][2]

    return None


def tabuleiro_cheio(tabuleiro):
    return all(tabuleiro[i][j] != ' ' for i in range(3) for j in range(3))


def ler_posicao(campo, jogador):
    while True:
        entrada = input(f"  Jogador {jogador} — {campo} (1 a 3): ").strip()
        try:
            valor = int(entrada)
            if 1 <= valor <= 3:
                return valor - 1
            print("  Valor inválido. Digite 1, 2 ou 3.\n")
        except ValueError:
            print("  Entrada inválida. Digite apenas um número inteiro.\n")


def main():
    tabuleiro = criar_tabuleiro()
    jogadores = ['X', 'O']
    turno = 0

    print("JOGO DA VELHA")

    print("Jogador 1 = X   |   Jogador 2 = O")


    exibir_tabuleiro(tabuleiro)

    while True:
        jogador = jogadores[turno % 2]
        print(f"Vez do Jogador {turno % 2 + 1} ({jogador})")

        while True:
            linha = ler_posicao("Linha ", jogador)
            col   = ler_posicao("Coluna", jogador)

            if tabuleiro[linha][col] != ' ':
                print("Posição já ocupada! Escolha outra.\n")
            else:
                break

        tabuleiro[linha][col] = jogador

        exibir_tabuleiro(tabuleiro)

        vencedor = verificar_vencedor(tabuleiro)
        if vencedor:
            print(f"Jogador {'1' if vencedor == 'X' else '2'} ({vencedor}) VENCEU!\n")
            break

        if tabuleiro_cheio(tabuleiro):
            print("Empate! Nenhum jogador venceu.\n")
            break

        turno += 1

    print("Fim de jogo.")
main()
```

```python
#exercício 3
def ler_medidas():
    while True:
        entrada = input("Quantas peças serão medidas? ").strip()
        try:
            n = int(entrada)
            if n > 0:
                return n
            print("  Digite um número maior que zero.\n")
        except ValueError:
            print("  Entrada inválida. Digite um número inteiro.\n")


def ler_float(mensagem):
    while True:
        entrada = input(mensagem).strip().replace(",", ".")
        try:
            valor = float(entrada)
            if valor >= 0:
                return valor
            print("  Digite um valor maior ou igual a zero.\n")
        except ValueError:
            print("  Entrada inválida. Digite um número.\n")


def filtrar_medidas(vetor, medida_desejada, tolerancia):
    limite_min = medida_desejada - tolerancia
    limite_max = medida_desejada + tolerancia
    validas = []
    for medida in vetor:
        if limite_min <= medida <= limite_max:
            validas.append(medida)
    return validas, limite_min, limite_max


def formatar_vetor(vetor):
    valores = []
    for v in vetor:
        if v == int(v):
            valores.append(str(int(v)))
        else:
            valores.append(str(v))
    return "{" + ", ".join(valores) + "}"


def main():
    print("   VERIFICAÇÃO DE MEDIDAS DE PEÇAS")

    n = ler_medidas()
    vetor = []
    print()
    for i in range(n):
        medida = ler_float(f"  Medida da peça {i + 1} (mm): ")
        vetor.append(medida)

    print()
    medida_desejada = ler_float("Medida desejável (mm): ")
    tolerancia      = ler_float("Tolerância (mm):       ")

    validas, limite_min, limite_max = filtrar_medidas(vetor, medida_desejada, tolerancia)

    print("  RESULTADO:")
    print(f"  Vetor de medições:      {formatar_vetor(vetor)} mm")
    print(f"  Medida desejável:       {medida_desejada} mm")
    print(f"  Tolerância:             ± {tolerancia} mm")
    print(f"  Intervalo válido:       [{limite_min} mm ; {limite_max} mm]")
    print(f"  Medições válidas:       {formatar_vetor(validas)} mm")
    print(f"  Quantidade de válidas:  {len(validas)}")
main()
```

```python
#exercício 4
def ler_gabarito():
    print("\nDigite o gabarito (10 questões, respostas de 'a' a 'e'):")



    gabarito = []
    respostas_validas = {'a', 'b', 'c', 'd', 'e'}

    for i in range(1, 11):
        while True:
            resposta = input(f"Questão {i}: ").strip().lower()
            if resposta in respostas_validas:
                gabarito.append(resposta)
                break
            else:
                print("  Resposta inválida! Digite apenas a, b, c, d ou e.")
    return gabarito


def ler_matricula():
    """Lê e valida a matrícula do aluno (número inteiro positivo)."""
    while True:
        try:
            matricula = int(input("Matrícula: "))
            if matricula > 0:
                return matricula
            else:
                print("  Matrícula inválida! Digite um número inteiro positivo.")
        except ValueError:
            print("  Matrícula inválida! Digite apenas números.")


def ler_respostas_aluno():
    """Lê e valida as 10 respostas de um aluno."""
    respostas = []
    respostas_validas = {'a', 'b', 'c', 'd', 'e'}

    print("Digite as respostas do aluno:")
    for i in range(1, 11):
        while True:
            resposta = input(f"  Questão {i}: ").strip().lower()
            if resposta in respostas_validas:
                respostas.append(resposta)
                break
            else:
                print("    Resposta inválida! Digite apenas a, b, c, d ou e.")
    return respostas

def calcular_nota(gabarito, respostas):
    """Calcula a nota comparando respostas com o gabarito."""
    acertos = sum(1 for g, r in zip(gabarito, respostas) if g == r)
    return acertos


def main():
    NUM_ALUNOS = 3
    MEDIA_APROVACAO = 7.0

    gabarito = ler_gabarito()

    alunos = []

    print("CADASTRO DOS ALUNOS")

    for i in range(1, NUM_ALUNOS + 1):
        print(f"\n--- Aluno {i} ---")
        matricula = ler_matricula()
        respostas = ler_respostas_aluno()
        nota = calcular_nota(gabarito, respostas)

        alunos.append({
            'matricula': matricula,
            'respostas': respostas,
            'nota': nota
        })

    print("RESULTADOS")
    print(f"\nGabarito: {' '.join(gabarito)}")

    aprovados = 0

    for aluno in alunos:
        print(f"\nMatrícula: {aluno['matricula']} ")
        print(f"Respostas: {' '.join(aluno['respostas'])}")
        print(f"Nota: {aluno['nota']:.1f}")

        if aluno['nota'] >= MEDIA_APROVACAO:
            print("Situação: APROVADO")
            aprovados += 1
        else:
            print("Situação: REPROVADO")

    percentual_aprovacao = (aprovados / NUM_ALUNOS) * 100

    print("ESTATÍSTICAS DA TURMA:")
    print(f"Total de alunos: {NUM_ALUNOS}")
    print(f"Aprovados: {aprovados}")
    print(f"Reprovados: {NUM_ALUNOS - aprovados}")
    print(f"Percentual de aprovação: {percentual_aprovacao:.2f}%")

main()
```

```python
#exercício 5
def converter24para12(horas, minutos):
    if horas == 0:
        hora12 = 12
        periodo = 'A'
    elif horas < 12:
        hora12 = horas
        periodo = 'A'
    elif horas == 12:
        hora12 = 12
        periodo = 'P'
    else:
        hora12 = horas - 12
        periodo = 'P'
    return hora12, periodo


def exibir_resultado(hora12, minutos, periodo):
    sufixo = "A.M." if periodo == 'A' else "P.M."
    print(f"\nResultado: {hora12}:{minutos:02d} {sufixo}")


def ler_inteiro(mensagem, minimo, maximo):
    while True:
        entrada = input(mensagem).strip()

        if entrada == "":
            print(f"Erro: campo obrigatorio. Digite um numero entre {minimo} e {maximo}.")
            continue

        if " " in entrada or "\t" in entrada:
            print("Erro: nao sao permitidos espacos.")
            continue

        if not entrada.lstrip('-').isdigit():
            print("Erro: apenas digitos sao permitidos.")
            continue

        if len(entrada.lstrip('-')) > 2:
            print("Erro: maximo de 2 digitos.")
            continue

        try:
            valor = int(entrada)
        except ValueError:
            print("Erro: valor nao e um inteiro valido.")
            continue

        if valor != float(entrada):
            print("Erro: nao sao permitidos valores decimais.")
            continue

        if valor < minimo:
            print(f"Erro: valor minimo e {minimo}.")
            continue

        if valor > maximo:
            print(f"Erro: valor maximo e {maximo}.")
            continue
        return valor


def main():
    while True:
        print("Digite o horario no formato 24h:")
        horas   = ler_inteiro("Horas   (0-23): ", 0, 23)
        minutos = ler_inteiro("Minutos (0-59): ", 0, 59)

        hora12, periodo = converter24para12(horas, minutos)
        exibir_resultado(hora12, minutos, periodo)

        print()
        resposta = input("Deseja converter outro horario? (s/n): ").strip().lower()
        while resposta not in ('s', 'n'):
            resposta = input("Digite 's' para sim ou 'n' para nao: ").strip().lower()

        if resposta == 'n':
            print("\nEncerrando o programa. Ate mais!")
            break

        print()
main()
```
